## 1. Install or locate Julia
The archive is downloaded from Julia's official S3 distribution if the executable is not already available.

```python
def detected_julia_version(executable):
    try:
        return subprocess.check_output([str(executable), '--startup-file=no', '-e', 'print(VERSION)'], text=True).strip()
    except (OSError, subprocess.CalledProcessError):
        return None

def locate_or_install_julia(version=JULIA_VERSION):
    existing = shutil.which('julia')
    if existing and detected_julia_version(existing) == version:
        return Path(existing)
    if existing:
        print(f'Ignoring system Julia {detected_julia_version(existing)} at {existing}; reproducibility requires {version}')
    install_root = WORK / f'julia-{version}'
    executable = install_root / 'bin' / 'julia'
    if executable.exists() and detected_julia_version(executable) == version:
        return executable
    minor = '.'.join(version.split('.')[:2])
    archive = WORK / f'julia-{version}-linux-x86_64.tar.gz'
    url = f'https://julialang-s3.julialang.org/bin/linux/x64/{minor}/julia-{version}-linux-x86_64.tar.gz'
    print('Downloading', url)
    urllib.request.urlretrieve(url, archive)
    with tarfile.open(archive, 'r:gz') as bundle:
        bundle.extractall(WORK)
    assert executable.exists(), f'Julia executable not found at {executable}'
    return executable

JULIA = locate_or_install_julia()
print('Julia:', JULIA)
actual_julia_version = detected_julia_version(JULIA)
print('Julia version:', actual_julia_version)
assert actual_julia_version == JULIA_VERSION, (actual_julia_version, JULIA_VERSION)
```

## 3. Instantiate the Julia environment
The first run can take several minutes because Julia compiles packages. Every run receives a separate timestamped directory under `/kaggle/working/sage_reproduction_logs`, preventing stale logs from being mixed into a new evidence bundle.

```python
BASE_ENV = os.environ.copy()
BASE_ENV['JULIA_DEPOT_PATH'] = str(DEPOT)
BASE_ENV['JULIA_NUM_THREADS'] = str(max(1, os.cpu_count() or 1))
BASE_ENV['OPENBLAS_NUM_THREADS'] = '1'
BASE_ENV['OMP_NUM_THREADS'] = '1'

def run_stage(name, args, required_token=None):
    command = [str(JULIA), '--project=.', *args]
    print('\n$', ' '.join(command), flush=True)
    started = time.perf_counter()
    process = subprocess.Popen(command, cwd=REPO, env=BASE_ENV, text=True, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, bufsize=1)
    lines = []
    with (LOG_DIR / f'{name}.log').open('w', encoding='utf-8') as live_log:
        for line in process.stdout:
            print(line, end='', flush=True)
            live_log.write(line)
            live_log.flush()
            lines.append(line)
    code = process.wait()
    elapsed = time.perf_counter() - started
    output = ''.join(lines)
    (LOG_DIR / f'{name}.log').write_text(output, encoding='utf-8')
    token_ok = required_token is None or required_token in output
    STAGE_RESULTS[name] = {'exit_code': code, 'seconds': round(elapsed, 3), 'required_token': required_token, 'token_ok': token_ok}
    (LOG_DIR / 'stage_results.json').write_text(json.dumps(STAGE_RESULTS, indent=2), encoding='utf-8')
    shutil.make_archive(str(WORK / 'sage_reproduction_bundle'), 'zip', root_dir=LOG_DIR)
    print(f'[{name}] exit={code} elapsed={elapsed:.3f}s token_ok={token_ok}', flush=True)
    if code != 0:
        raise RuntimeError(f'{name} failed with exit code {code}')
    if not token_ok:
        raise AssertionError(f'{name} did not emit {required_token!r}')
    return output

instantiate_log = run_stage('00_instantiate', ['-e', 'using Pkg; Pkg.instantiate(); Pkg.precompile()'])
```

## 19. Phase 18 external literature-family baselines
Protocol v3: each of six methods gets nine validation configurations, two development seeds and 128 updates per configuration (108 validation training runs total). DBB tunes lambda in {0.05, 0.5, 5}; DPO tunes Gaussian RMS multipliers in {0.5, 1, 2}; compact smoothing/SAGE tune fixed-per-run rho in {0.25, 0.5, 1}. These four methods cross their scale grid with learning rates {0.0003, 0.001, 0.003}. STE/hard-local instead get nine learning rates across the same range. Samples remain N=4. Equal candidate/update budgets do not imply equal tuning CPU cost; validation work is reported separately. Test pools stay unopened until selection is frozen. Preflight profiles central and extreme configurations, checks real sample threads and serial/resume parity, and estimates each checkpointed stage with a 2x margin. This ETA is not a runtime guarantee. Prepare performs validation-only tuning and a 256-update validation pilot. Held-out confirmation is split into fixed shards 11-12 and 13-15; merge refuses to report confidence intervals until all 45 rows exist. Confirmation uses 1024 fixed updates and actual-time replay controls for STE/DBB/DPO; negative scientific results remain COMPLETE. Checkpoints are atomic every 16 updates.

All methods share the same MLP and exact center projection P(z). DBB uses [P(z)-P(z-lambda*v)]/lambda with upstream v. Gaussian VJP uses the sample average of epsilon * dot(P(z+sigma*epsilon)-P(z), v)/sigma, with sigma = multiplier*rho/sqrt(d+6). This Gaussian VJP smooths the optimizer map with a frozen upstream, rather than the composed task loss used by compact loss smoothing/SAGE. DBB and DPO here are explicit quadratic-projection / hard-forward adaptations of [Vlastelica et al. (ICLR 2020)](https://arxiv.org/abs/1912.02175) and [Berthet et al. (NeurIPS 2020)](https://arxiv.org/abs/2002.08676), not reproductions of their original experiments or an established SOTA comparison.

The dataset is controlled synthetic 3x4 job-shop, not an external real-world dataset. Development seeds 1-2 use a separate IID validation pool. Confirmation uses training seeds 11-15 with new 32-instance IID and 32-instance OOD pools, shared across methods. Primary SAGE-minus-DBB/DPO comparisons use paired seed intervals with Bonferroni adjustment for the two claims; time-budget replays are exploratory and report overshoot.
